```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline

print("Libraries imported and ready to use")
```

```python
# Load the dataset

df = pd.read_csv('C:\\Users\\yashw\\Documents\\Projects\\Diamond-Price-Prediction\\Data\\raw\\train.csv')
print("Data loaded successfully")
print("Shape:", df.shape)
```

```python
# checking for unique values in categorical columns

print("Cut categories:", df['cut'].unique())
print("Color categories:", df['color'].unique())
print("Clarity categories:", df['clarity'].unique())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Price vs Count 

plt.figure(figsize=(10, 5))
sns.histplot(df['price'], bins=50, kde=True, color='blue')
plt.title("Distribution of Diamond Price")
plt.xlabel("Price")
plt.ylabel("Count")
plt.show()
```

```python
# Price vs Carat

plt.figure(figsize=(10, 5))
sns.scatterplot(x='carat', y='price', data=df, alpha=0.3, color='green')
plt.title("Carat vs Price")
plt.xlabel("Carat")
plt.ylabel("Price")
plt.show()
```

```python
# Correlation heatmap for numerical features

plt.figure(figsize=(10, 8))
numerical_df = df.select_dtypes(include=['float64', 'int64'])
sns.heatmap(numerical_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Checking correlation for the Cut feature which is categorical column

plt.figure(figsize=(10, 5))
sns.boxplot(x='cut', y='price', data=df, 
            order=['Fair', 'Good', 'Very Good', 'Premium', 'Ideal'])
plt.title("Price by Cut Quality")
plt.xlabel("Cut")
plt.ylabel("Price")
plt.show()
```

```python
# Now correlation for Color feature

plt.figure(figsize=(10, 5))
sns.boxplot(x='color', y='price', data=df,
            order=['D', 'E', 'F', 'G', 'H', 'I', 'J'])
plt.title("Price by Color")
plt.xlabel("Color")
plt.ylabel("Price")
plt.show()
```
